## Generate code

Generate code for data preprocessing, including handling categorical features and scaling numerical features.


**Reasoning**:
Import necessary libraries and perform data preprocessing steps as outlined in the instructions, including one-hot encoding categorical features and standard scaling numerical features.



```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identify categorical columns
categorical_cols = ['Location', 'PaymentMethod']

# Instantiate and fit_transform OneHotEncoder
onehot_encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoded_categorical_features = onehot_encoder.fit_transform(df[categorical_cols])

# Create a DataFrame from the one-hot encoded features
encoded_categorical_df = pd.DataFrame(encoded_categorical_features, columns=onehot_encoder.get_feature_names_out(categorical_cols))

# Identify numerical columns (excluding 'Timestamp', 'CustomerID', 'TransactionID')
numerical_cols = ['Amount']

# Instantiate and fit_transform StandardScaler
scaler = StandardScaler()
scaled_numerical_features = scaler.fit_transform(df[numerical_cols])

# Create a DataFrame from the scaled numerical features
scaled_numerical_df = pd.DataFrame(scaled_numerical_features, columns=numerical_cols)

# Combine the encoded categorical features and scaled numerical features
preprocessed_df = pd.concat([encoded_categorical_df, scaled_numerical_df], axis=1)

# Display the first few rows of the preprocessed DataFrame
print("First few rows of the preprocessed data:")
display(preprocessed_df.head())

# Display info about the preprocessed DataFrame
print("\nInfo about the preprocessed data:")
preprocessed_df.info()
```
